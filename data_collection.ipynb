```python
import os
import pandas as pd
from sodapy import Socrata

# Credenciales de datos.gov.co
DOMAIN = "www.datos.gov.co"
DATASET_ID = "p6dx-8zbt"  # SECOP II - Procesos de Contratación

# Coloca tu App Token aquí
APP_TOKEN = "qhefb31lPp4tliAYftKFzHOlp" 

# Inicializar cliente pasando solo el app_token
client = Socrata(
    DOMAIN,
    APP_TOKEN
)

client.timeout = 120
```

```python
# 2. Definición de la consulta SoQL (Filtro espacial y temporal)
# Filtramos por entidades del Valle del Cauca desde el 2024 en adelante
where_clause = (
    "departamento_entidad = 'Valle del Cauca' AND "
    "fecha_de_publicacion_del >= '2024-01-01T00:00:00.000'"
)

# Columnas específicas relevantes para el Clustering
select_columns = (
    "id_del_proceso, referencia_del_proceso, entidad, nit_entidad, ciudad_entidad, "
    "ordenentidad, modalidad_de_contratacion, tipo_de_contrato, estado_resumen, "
    "precio_base, valor_total_adjudicacion, duracion, unidad_de_duracion, "
    "proveedores_invitados, respuestas_al_procedimiento, proveedores_unicos_con, "
    "fecha_de_publicacion_del"
)

# 3. Proceso de descarga paginada por lotes
limit = 5000
offset = 0
all_records = []

print("Iniciando la descarga de datos desde datos.gov.co...")

while True:
    results = client.get(
        DATASET_ID,
        select=select_columns,
        where=where_clause,
        limit=limit,
        offset=offset,
        order="id_del_proceso ASC"
    )
    
    if not results:
        break
        
    all_records.extend(results)
    print(f"Descargados {len(all_records)} registros...")
    offset += limit

# 4. Conversión a DataFrame de Pandas
df_secop = pd.DataFrame.from_dict(all_records)
print(f"\n¡Descarga completada con éxito!")
print(f"Total de registros cargados: {df_secop.shape[0]} | Columnas: {df_secop.shape[1]}")
```

```python
# Guardar dataset localmente como CSV
df_secop.to_csv("secop2_valle_2024.csv", index=False, encoding="utf-8")
print("¡Archivo guardado con éxito como 'secop2_valle_2024.csv'!")
```
